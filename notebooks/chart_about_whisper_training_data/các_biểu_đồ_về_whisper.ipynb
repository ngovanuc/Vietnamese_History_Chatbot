```python
import matplotlib.pyplot as plt

# Tên cột
labels = ['English', 'Non-English', 'Total']

# Dữ liệu cột đơn
english_hours = 563
non_english_hours = 117

# Dữ liệu cột stack
evaluate_hours = 125
training_hours = 680 - evaluate_hours  # 680 tổng - 125 evaluate

# Tạo dữ liệu stack: 0 cho các cột không có phần đó
evaluate = [0, 0, evaluate_hours]
training = [0, 0, training_hours]

# Thiết lập biểu đồ
fig, ax = plt.subplots(figsize=(8, 6))

# Vẽ cột đơn
ax.bar(labels[0], english_hours, label='English', color='skyblue')
ax.bar(labels[1], non_english_hours, label='Non-English', color='lightgreen')

# Vẽ cột stack với màu rõ ràng hơn
ax.bar(labels[2], training[2], label='Training', color='#1f77b4')     # xanh dương đậm
ax.bar(labels[2], evaluate[2], bottom=training[2], label='Evaluate', color='#d62728')  # đỏ tươi

# Nhãn và tiêu đề
ax.set_ylabel('Giờ (nghìn)')
ax.set_title('Phân bố giờ theo ngôn ngữ và mục đích sử dụng')
ax.set_ylim(0, 750)
ax.legend()

# Thêm giá trị trên mỗi cột
for i, (x, y) in enumerate(zip(labels, [english_hours, non_english_hours, 680])):
    ax.text(i, y + 10, f'{y}K', ha='center', fontsize=10)

# Hiển thị
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Dữ liệu
labels = ['English', 'Non-English']
sizes = [563, 117]
colors = ['skyblue', 'lightgreen']
explode = (0.05, 0)  # Tách nhẹ English ra khỏi tâm

# Vẽ biểu đồ tròn
fig, ax = plt.subplots(figsize=(6, 6))
ax.pie(
    sizes,
    labels=labels,
    colors=colors,
    autopct='%1.1f%%',
    startangle=90,
    explode=explode,
    shadow=True
)

# Thiết lập dạng tròn và tiêu đề
ax.axis('equal')  # Đảm bảo hình tròn đều
plt.title('Tỷ lệ giờ theo ngôn ngữ (English vs Non-English)', fontsize=14)

# Hiển thị
plt.show()
```

```python
import matplotlib.pyplot as plt

# Dữ liệu
sizes = [500, 10, 30]
colors = ['#66b3ff', '#99ff99', '#ff9999']
explode = (0.05, 0, 0)

# Tạo biểu đồ lớn hơn (gấp rưỡi)
fig, ax = plt.subplots(figsize=(10, 10))

# Vẽ biểu đồ tròn, chỉ hiển thị phần trăm
wedges, texts, autotexts = ax.pie(
    sizes,
    labels=[''] * 3,
    autopct=lambda pct: f"{pct:.1f}%",  # chỉ phần trăm
    startangle=140,
    colors=colors,
    explode=explode,
    textprops={'fontsize': 12},
    wedgeprops=dict(edgecolor='w')
)

# Legend đầy đủ tên và số giờ
full_labels = [
    'linhtran92/viet_bud500 (~500 giờ)',
    'CodeLinkIO/vietnamese-voice-dataset (~10 giờ)',
    'FOSD - FPT Corporation (~30 giờ)'
]
ax.legend(wedges, full_labels, title='Nguồn dữ liệu', loc='center left', bbox_to_anchor=(1, 0.5), fontsize=13)

# Tiêu đề và hiển thị
plt.title('Phân bố giờ theo nguồn dữ liệu', fontsize=16)
plt.tight_layout()
plt.show()
```
