# Connect mongodb

```python
import pymongo
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")
# client = MongoClient("mongodb://<or you can replace your favorite connection name here>")
# like: client = MongoClient("mongodb:MY_CONNECTION")
```

# Tạo database và collection

```python
db = client.test_database
test_collection = db.test_collection
```

# Chèn dữ liệu vào vector database

```python
import datetime

post = {
    "author":  "Mike",
    "text":  "My first blog post!",
    "tags": ["mongodb", "python", "pymongo"],
    "datetime": datetime.datetime.now(tz=datetime.timezone.utc)
}
```

```python
# Insert documents
posts = db.posts
post_id = posts.insert_one(post).inserted_id
post_id
```

```python
test_collection.insert_one(post).inserted_id
```

# Xóa một collection

```python
db.drop_collection("posts")
```

# In ra toàn bộ dữ liệu trong một collection

```python
# print all documents in the collection
for post in test_collection.find():
    print(post)
```

# Thêm một trường (cột) mới vào document

```python
# add a new field to the document
post["author"] = "John Doe"
posts.update_one({"_id": post_id}, {"$set": post})
# find the document by its ID
post = posts.find_one({"_id": post_id})
print(post)
```

# Sử dụng lại một database đã được tạo trước đó

```python
from pymongo import MongoClient

uri = "mongodb://localhost:27017/"
client = MongoClient(uri)

database = client.get_database("test_database")
collection = database.get_collection("test_collection")

query = {"name": "Mike"}
author = collection.find_one(query)
print(author)

client.close()
```
